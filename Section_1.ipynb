## **INTRODUCTION**
### **Psi4 Installation**

This section in the tutorial just details the installation process of psi4.

Instructions for downloading with different operating systems are detailed on Psi4's official download page at [https://psicode.org/installs/v111/](https://psicode.org/installs/v111/).

For WSL users the installation goes as follows (***on the Ubuntu terminal***):

 **1. Download the installer package**
curl "https://vergil.chemistry.gatech.edu/psicode-download/Psi4conda-1.11-py314-Linux-x86_64.sh" -o Psi4conda-1.11-py314-Linux-x86_64.sh --keepalive-time 2

**2. Run the installer script in batch mode (-b) to target your home directory (-p)**
bash Psi4conda-1.11-py314-Linux-x86_64.sh -b -p $HOME/psi4conda

**3. Append the environment activation hook directly to your Ubuntu bash configuration**
echo $'. $HOME/psi4conda/etc/profile.d/conda.sh\nconda activate' >> ~/.bashrc

**4. Reload your shell file immediately (instead of logging out and back in)**
source ~/.bashrc
